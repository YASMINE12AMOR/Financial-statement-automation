```python
def split_pdf(input_pdf_path, output_folder):
    # Open the PDF file
    with open(input_pdf_path, 'rb') as file:
        pdf_reader = PyPDF2.PdfReader(file)
        num_pages = len(pdf_reader.pages)

        # Iterate through each page in the PDF
        for page_num in range(num_pages):
            # Create a new PDF writer for each page
            writer = PdfWriter()

            writer.add_page(pdf_reader.pages[page_num])

            # Output file name for each page
            output_filename = f"{output_folder}/page_{page_num + 1}.pdf"

            # Write the page to a new PDF file
            with open(output_filename, 'wb') as output_file:
                writer.write(output_file)

# Example usage:
input_pdf_path = "/content/cnt_efc311217.pdf"  # Change this to your input PDF file
output_folder = "/content/drive/MyDrive/bunch"  # Change this to the folder where you want to save the individual PDFs
split_pdf(input_pdf_path, output_folder)
```

```python
def intersection(box_1, box_2):
  return [box_2[0], box_1[1],box_2[2], box_1[3]]
```

```python
def iou(box_1, box_2):

  x_1 = max(box_1[0], box_2[0])
  y_1 = max(box_1[1], box_2[1])
  x_2 = min(box_1[2], box_2[2])
  y_2 = min(box_1[3], box_2[3])

  inter = abs(max((x_2 - x_1, 0)) * max((y_2 - y_1), 0))
  if inter == 0:
      return 0

  box_1_area = abs((box_1[2] - box_1[0]) * (box_1[3] - box_1[1]))
  box_2_area = abs((box_2[2] - box_2[0]) * (box_2[3] - box_2[1]))

  return inter / float(box_1_area + box_2_area - inter)
```

```python
import spacy
from sklearn.metrics.pairwise import cosine_similarity
from collections import defaultdict
import numpy as np
```

```python
# Tokenize the OCR output using spaCy
tokens = []
for row in new_out_array:
    tokens.extend([token.text for token in nlp(' '.join(row))])

# Initialize a dictionary to store word vectors
word_vectors = defaultdict()

# Compute word embeddings and store them in the dictionary
for token in tokens:
    if token not in word_vectors:
        word_vectors[token] = nlp(token).vector

# Define a similarity threshold (adjust as needed)
similarity_threshold = 0.9

# Remove duplicates based on similarity
unique_tokens = []
seen = set()
for token in tokens:
    if token not in seen:
        unique_tokens.append(token)
        seen.add(token)
    else:
        # Check similarity with previously seen tokens
        is_duplicate = False
        for seen_token in seen:
            similarity = cosine_similarity([word_vectors[token]], [word_vectors[seen_token]])[0][0]
            if similarity > similarity_threshold:
                is_duplicate = True
                break
        if not is_duplicate:
            unique_tokens.append(token)
            seen.add(token)

# Output the unique tokens
unique_text = ' '.join(unique_tokens)
print(unique_text)
```

```python
# Initialiser un ensemble pour stocker les phrases déjà vues
seen_phrases = set()

# Parcourir chaque sous-liste dans la liste de listes
for row in out_array:
    # Initialiser une liste temporaire pour stocker les phrases uniques dans cette sous-liste
    unique_row = []
    # Parcourir chaque phrase dans la sous-liste actuelle
    for phrase in row:
        # Nettoyer la phrase en enlevant les espaces inutiles et en la mettant en minuscules
        cleaned_phrase = phrase.strip().lower()
        # Vérifier si la phrase n'a pas déjà été vue
        if cleaned_phrase not in seen_phrases:
            # Ajouter la phrase à l'ensemble des phrases vues
            seen_phrases.add(cleaned_phrase)
            # Ajouter la phrase à la liste temporaire des phrases uniques
            unique_row.append(phrase)
        else:
            # Si la phrase a déjà été vue, la laisser vide
            unique_row.append('')
    # Remplacer la sous-liste actuelle par la liste temporaire des phrases uniques
    row[:] = unique_row

# Afficher le résultat
for row in out_array:
    print(row)
```

```python
import PyPDF2
from pdf2image import convert_from_path
import cv2
import numpy as np
import pytesseract
import io
import os
import re  # Import du module re pour les expressions régulières
```

```python

def has_notes(text):
    exclude_keywords = ['bilan', 'bilan consolidé',
                        'total actif', 'total passif', 'total des actif', 'total des passif',
                        "total de l'actif", "total des passifs", 'total actifs', 'total passifs', 'total des actifs',
                        "hors bilan",'etat de resultat','etats de resultat','etat de résultat',"etat de flux de trésorerie",
                        "etat de flux de tresorerie", "etat de variation de l'actif net trimestriel","etats de variation de l'actif net",
                        "etat des flux de trésorerie", "etat des flux de tresorerie"]


    for total_keyword in extracted_list:
        if re.search(r'\b{}\b'.format(total_keyword), text, re.IGNORECASE):
            if not any(re.search(r'\b{}\b'.format(exclude_keyword), text, re.IGNORECASE) for exclude_keyword in exclude_keywords):
               return True
    return False

def find_and_save_note_data(pdf_file_path, output_file_path):
    try:
        print("Processing:", pdf_file_path)
        found_bilan = False
        writer = PyPDF2.PdfWriter()

        with open(pdf_file_path, 'rb') as file:
            reader = PyPDF2.PdfReader(file)
            for page_num, page in enumerate(reader.pages):
                text = page.extract_text()
                if has_notes(text):
                    found_bilan = True
                    writer.add_page(page)

        # If bilan is not found using text extraction, try OCR for each page
        if not found_bilan:
            print("note not found using text extraction. Trying OCR.")
            images = convert_from_path(pdf_file_path)

            for img_num, img in enumerate(images):
                gray = cv2.cvtColor(np.array(img), cv2.COLOR_BGR2GRAY)
                extracted_text = pytesseract.image_to_string(gray)

                if has_notes(extracted_text):
                    found_bilan = True
                    img_bytes = io.BytesIO()
                    img.save(img_bytes, format='PDF')
                    writer.add_page(PyPDF2.PdfReader(img_bytes).pages[0])

        if found_bilan:
            with open(output_file_path, 'wb') as output_file:
                writer.write(output_file)
            print("notes data saved to:", output_file_path)
        else:
            print("No 'notes' found in the PDF:", pdf_file_path)

    except Exception as e:
        print("An error occurred while processing the PDF file:", e)

input_dir = "/content/drive/MyDrive/new"  # Chemin du répertoire contenant les fichiers PDF
output_dir = "/content/drive/MyDrive/output/notes"

if not os.path.exists(output_dir):
    os.makedirs(output_dir)

for filename in os.listdir(input_dir):
    if filename.endswith('.pdf'):
        pdf_file_path = os.path.join(input_dir, filename)
        output_filename = os.path.basename(pdf_file_path)
        output_file_path = os.path.join(output_dir, output_filename)
        find_and_save_note_data(pdf_file_path, output_file_path)
```

```python
from PIL import Image, ImageEnhance

#read the image
im = Image.open("ext_im.jpg")

#image brightness enhancer
enhancer = ImageEnhance.Brightness(im)

factor = 1 #gives original image
im_output = enhancer.enhance(factor)
im_output.save('ext_im-1.jpg')

factor = 1.5## brightens the image
im_output = enhancer.enhance(factor)
im_output.save('ext_im-2.jpg')
```
